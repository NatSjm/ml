```python
# 1. Median before
median_before = df['horsepower'].median()
print('median before:', median_before)
```

```python
# 2. Most frequent value (the "mode")
most_frequent = df['horsepower'].mode()[0]
print('most frequent:', most_frequent)
```

```python
hp_filled = df['horsepower'].fillna(most_frequent)
```

```python
median_after = hp_filled.median()
print('median after:', median_after)
```

```python
# 1-3. Cars from Asia, two columns, first 7 rows
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
```

```python
# 4. Convert to a NumPy array
X = X.values
X
```

```python
# 5. X transposed times X
XTX = X.T.dot(X)
```

```python
# 6. Invert it
XTX_inv = np.linalg.inv(XTX)
```

```python
# 7. The y values
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
# 9. Sum of the elements of w
w.sum()
```
